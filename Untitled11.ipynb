```python
import zipfile
import os
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, TensorDataset

zip_path = "/content/lode_runner.zip"
extract_path = "/content/lode_runner"

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(extract_path)

level_folder = "/content/lode_runner/Lode Runner/Processed"

files = os.listdir(level_folder)

print("Number of files:", len(files))
print(files[:10])
```

```python
import matplotlib.pyplot as plt
```

```python
levels = []

for filename in sorted(os.listdir(level_folder)):

    if filename.endswith(".txt"):

        file_path = os.path.join(level_folder, filename)

        with open(file_path, "r") as file:
            level = file.read().splitlines()

        levels.append(level)

print("Levels loaded:", len(levels))

for row in levels[0]:
    print(row)
```

```python
for i, level in enumerate(levels):

    rows = len(level)
    columns = len(level[0])

    print(f"Level {i+1}: {rows} x {columns}")
```

```python
tile_types = set()

for level in levels:
    for row in level:
        for tile in row:
            tile_types.add(tile)

print("Tile types:", sorted(tile_types))
print("Number of tile types:", len(tile_types))
```

```python
from collections import Counter

tile_counts = Counter()

for level in levels:
    for row in level:
        tile_counts.update(row)

print(tile_counts)
```

```python
valid_tiles = {'.', 'b', '#', 'B', '-', 'G', 'E', 'M'}
invalid_tiles = set()

for level in levels:
    for row in level:
        for tile in row:
            if tile not in valid_tiles:
                invalid_tiles.add(tile)

if len(invalid_tiles) == 0:
    print("No invalid tiles found.")
else:
    print("Invalid tiles:", invalid_tiles)
```

```python
level_strings = []

for level in levels:
    level_string = "\n".join(level)
    level_strings.append(level_string)

unique_levels = set(level_strings)

print("Total levels:", len(level_strings))
print("The first five levels", level_strings[:5])
print("Unique levels:", len(unique_levels))
print("Duplicates:", len(level_strings) - len(unique_levels))
```

```python
tile_to_int = {'.': 0,'b': 1,'#': 2,'B': 3,'-': 4,'G': 5,'E': 6,'M': 7
}

encoded_levels = []

for level in levels:
    encoded_level = []
    for row in level:
        encoded_row = [tile_to_int[tile] for tile in row]
        encoded_level.append(encoded_row)

    encoded_levels.append(encoded_level)
encoded_levels = np.array(encoded_levels)

print("Encoded shape:", encoded_levels.shape)

num_tiles = len(tile_to_int)
one_hot_levels = np.eye(num_tiles)[encoded_levels]
print("One-hot shape:", one_hot_levels.shape)
```

```python
np.random.seed(42)
np.random.shuffle(one_hot_levels)
print("Shuffled:", one_hot_levels.shape)
```

```python
tot_lev = len(one_hot_levels)
train_split = int(0.70 * tot_lev)
val_split = int(0.15 * tot_lev)
test_split = tot_lev - train_split - val_split

print("Train:", train_split)
print("Val:", val_split)
print("Test:", test_split)
```

```python
train_data = one_hot_levels[:train_split]
val_data = one_hot_levels[train_split:train_split+val_split]
test_data = one_hot_levels[train_split+val_split:]
print("Train:", train_data.shape)
print("Val:", val_data.shape)
print("Test:", test_data.shape)
```

```python
print("Original:", len(levels))
print("Train:", len(train_data))
print("Val:", len(val_data))
print("Test:", len(test_data))
```

```python
dataset_train = TensorDataset(tensor_tr)
tr_loader = DataLoader(dataset_train, batch_size=16, shuffle=True)

print("Training batches:", len(tr_loader))
```

```python
#Generator
class Generator(nn.Module):
  def __init__(self, noise_dim = 100, num_tiles = 8):
    super().__init__()
    #converts random no in2 features
    self.fc = nn.Linear(noise_dim, 128 * 11 * 16)
    self.conv = nn.Sequential(
        nn.Conv2d(128, 64, kernel_size=3, padding= 1),
        nn.ReLU(),
        #learns the spatial patterns
        nn.Conv2d(64, num_tiles, kernel_size=3, padding = 1))

  def forward(self, noise):
    x = self.fc(noise)
    #reshapes features in2 grids
    x = x.view(-1, 128, 11, 16)
    #enlarges the grid
    x = F.interpolate(x, size = (22, 32), mode="nearest")
    x = self.conv(x)
    return F.softmax(x, dim = 1)
```

```python
#Discriminator
class Discriminator(nn.Module):
  def __init__(self, num_tiles = 8):
    super().__init__()
    #allows neural networks to be stacked
    self.model = nn.Sequential(
        #extracts pattern from level layout
        nn.Conv2d(num_tiles, 32, kernel_size=3, stride=2, padding=1),
        #network can learn more complex rela
        nn.LeakyReLU(0.2),
        #extracted features 2 single vector
        nn.Flatten(),
        #prod 1 score to see if the input looks real or generated
        nn.Linear(32 * 11 * 16, 1))

  def forward(self, x):
    return self.model(x)

# redefine
gen = Generator().to(dev)
disc = Discriminator().to(dev)

# fake levels 2 check if it works
gen_noise = torch.randn(4, 100, device=dev)
gen_flevels = gen(gen_noise)
print("Generator output shape:", gen_flevels.shape)

disc_score = disc(gen_flevels)
print("Discriminator output shape:", disc_score.shape)
```

```python
#train the GAN
#binary cross entropy less (discriminator is binary)
loss_f = nn.BCEWithLogitsLoss()
#opt algorithm, adjusts model wight based on calc gradients
optmizer_gen = optim.Adam(gen.parameters(), lr=0.0002, betas=(0.5, 0.999))
optimizer_disc = optim.Adam(disc.parameters(), lr=0.0002, betas=(0.5, 0.999))
```

```python

# Training settings
epochs = 100
gen_rand = 100

gen_loss = []
disc_loss = []

# Outer loop: repeat for 100 epochs
for epoch in range(epochs):

    gen.train()
    disc.train()

    ep_gen_loss = 0
    ep_disc_loss = 0

    # Inner loop: process all 7 batches each epoch
    for (real_levels,) in tr_loader:

        real_levels = real_levels.to(dev)
        batch_size = real_levels.size(0)

        # STEP 1: TRAIN DISCRIMINATOR
        optimizer_disc.zero_grad()

        # Real examples
        labels_r = torch.ones(batch_size, 1, device=dev)
        outp_r = disc(real_levels)
        loss_r = loss_f(outp_r, labels_r)

        # Fake examples
        gen_noise = torch.randn(batch_size, gen_rand, device=dev)
        fake_levels = gen(gen_noise)

        fake_lab = torch.zeros(batch_size, 1, device=dev)
        outp_f = disc(fake_levels.detach())
        loss_fake = loss_f(outp_f, fake_lab)

        # Update Discriminator
        loss_disc = loss_r + loss_fake
        loss_disc.backward()
        optimizer_disc.step()

        # STEP 2: TRAIN GENERATOR
        optmizer_gen.zero_grad()

        # Generate new fake levels
        gen_noise = torch.randn(batch_size, gen_rand, device=dev)
        fake_levels = gen(gen_noise)

        # Generator wants Discriminator to predict REAL
        outp_gen = disc(fake_levels)
        loss_gen = loss_f(outp_gen, labels_r)

        # Update Generator
        loss_gen.backward()
        optmizer_gen.step()

        # Record batch losses
        ep_gen_loss += loss_gen.item()
        ep_disc_loss += loss_disc.item()

    # These lines are inside the epoch loop,
    # but outside the batch loop
    gen_loss.append(ep_gen_loss / len(tr_loader))
    disc_loss.append(ep_disc_loss / len(tr_loader))

    # Print progress every 10 epochs
    if (epoch + 1) % 10 == 0:
        print(
            f"Epoch {epoch+1}/{epochs}, "
            f"Gen Loss: {gen_loss[-1]:.4f}, "
            f"Disc Loss: {disc_loss[-1]:.4f}"
        )

print("Generator losses recorded:", len(gen_loss))
print("Discriminator losses recorded:", len(disc_loss))
```
